## Pandas data analysis

```python
import pandas as pd
```

```python
dataset_path = "../ai_jobs_market_2025_2026.csv"

df = pd.read_csv(dataset_path)

#check shape
print("Shape of dataset : rows X columns:", df.shape)
print("\nColumn names:\n", df.columns.tolist())
```

```python
# First and Last Rows
df.head(2)
#df.tail(2)
```

```python
print("\nInfo summary:", df.info())   # Column names, dtypes, and non-null counts
```

```python
# statistical ( Count, mean, std, min, max, percentiles) summary of all numeric columns
print("\nDescribe (numeric columns):")
print(df.describe())
```

```python

# Describe including all categorical/text columns
print("\nDescribe (all columns):")
print(df.describe(include="all"))
```

```python
#  Count missing values per column 
print("Missing values per column:")
print(df.isnull().sum())

# Total rows with atleast one missing value,  
rows_with_any_missing = df.isnull().any(axis=1).sum()
print(f"\nRows with at least one missing value: {rows_with_any_missing} ({rows_with_any_missing/len(df)*100:.1f}%)")
```

```python
# Check for duplicates
duplicate_count = df.duplicated().sum()
print(f"Fully duplicate rows: {duplicate_count}")
```

```python
# check if any row has salary_min_usd greater than annual_salary_usd
invalid_min = df[df["salary_min_usd"] > df["annual_salary_usd"]]
print("Rows where salary_min > annual_salary:", len(invalid_min))

invalid_max = df[df["annual_salary_usd"] > df["salary_max_usd"]]
print("Rows where annual_salary > salary_max:", len(invalid_max))

invalid_max = df[df["annual_salary_usd"] > df["salary_max_usd"]] # list few salaries > sallary_max
print(invalid_max[["job_id", "annual_salary_usd", "salary_min_usd", "salary_max_usd"]].head(10))
```

```python
# check rows where the actual salary paid is higher than the stated maximum salary
invalid_max = df[df["annual_salary_usd"] > df["salary_max_usd"]]
invalid_max[["job_id", "annual_salary_usd", "salary_min_usd", "salary_max_usd"]].head(10)
```

```python
# calculate how much % the annual salary exceeds the stated maximum

invalid_max_copy = invalid_max.copy()  # avoid changing the original dataframe by accident
invalid_max_copy["pct_over_max"] = (
    (invalid_max_copy["annual_salary_usd"] - invalid_max_copy["salary_max_usd"])
    / invalid_max_copy["salary_max_usd"] * 100
)

# look at the summary of how far over the max these rows go
invalid_max_copy["pct_over_max"].describe()
```

```python
# check: do "violating" rows have a higher AI salary premium on average?
violates_max = df["annual_salary_usd"] > df["salary_max_usd"]

# compare avg premium % for violators vs non-violators
df.groupby(violates_max)["ai_salary_premium_pct"].mean()
```

```python
# flag rows where salary is above the stated max
# keeps original data untouched, just marks the issue for later use
df["salary_exceeds_max"] = df["annual_salary_usd"] > df["salary_max_usd"]

# quick check: how many rows got flagged
df["salary_exceeds_max"].sum()
```

```python
# check for any zero or negative salary
(df[["annual_salary_usd", "salary_min_usd", "salary_max_usd"]] <= 0).sum()
```

```python
# see how required_skills looks before we split it
df["required_skills"].head(5)
```

```python
# split each skill by "|" and make it a list
df["skills_list"] = df["required_skills"].str.split("|", regex=False)

# check how it looks now
df["skills_list"].head(5)
```

```python
# count how many skills each job needs
df["num_skills"] = df["skills_list"].str.len()

# check a few rows
df["num_skills"].head(5)
```

```python
# quick check - min, max, avg number of skills across all jobs
df["num_skills"].describe()
```

```python
# check posting_year and posting_month before combining
df[["posting_year", "posting_month"]].head(5)
```

```python
# combine year and month into one real date column
df["posting_date"] = pd.to_datetime(
    df["posting_year"].astype("string") + "-" + df["posting_month"].astype("string") + "-01", errors="coerce")

# check the output
df["posting_date"].head(5)
```

```python
# check min and max date - to see if range looks correct
df["posting_date"].min(), df["posting_date"].max()
```

```python
# average salary for each experience level
df.groupby("experience_level")["annual_salary_usd"].mean()
```

```python
# get mean, min, max, count - for each experience level
df.groupby("experience_level")["annual_salary_usd"].agg(["mean", "min", "max", "count"])
```

```python
# compare avg salary for remote vs non-remote jobs
df.groupby("remote_work")["annual_salary_usd"].mean()
```

```python
# pivot table - avg salary by experience level and remote work type
df.pivot_table(
    values="annual_salary_usd",
    index="experience_level",
    columns="remote_work",
    aggfunc="mean"
)
```
