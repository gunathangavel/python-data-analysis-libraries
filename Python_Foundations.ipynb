# Python Core concepts

### A lambda function is an anonymous, one-line function used for short operations, often passed as arguments to functions like filter, map, and sorted
- syntax = lambda arguments: expression --> arguments = inputs; expression = output (only one expression allowed)

```python
#lambda

square = lambda x: x * x
print(square(5))  # 25

is_even = lambda x: x % 2 == 0
print(is_even(4))  # True
```
